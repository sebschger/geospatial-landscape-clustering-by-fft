```python
import matplotlib.pyplot as plt
import numpy as np
import pickle

with open("output/temporary_masks.pkl", "rb") as f:
    circle_masks = pickle.load(f)

print(circle_masks[5].shape)

def radius_divisor(shape):
    """Für jedes FFT-Pixel: Anzahl der Pixel auf demselben Radius."""
    ny, nx = shape
    cy, cx = ny // 2, nx // 2          # nach fftshift
    y, x = np.indices((ny, nx))
    r_int = np.sqrt((y - cy)**2 + (x - cx)**2).astype(int)
    counts = np.bincount(r_int.ravel()) # Pixel pro Radius
    return counts[r_int]                # Array in FFT-Form


map = radius_divisor((21,21))

plt.imshow(map)
plt.show()
```

```python
test_circle = (circle_masks / map)
```

```python
import matplotlib.pyplot as plt

for mask in test_circle:
    plt.imshow(mask)
    plt.show()
    print(mask.sum())
```
